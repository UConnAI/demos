```python
import cherche
from cherche import retrieve, rank
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
corpus_df = pd.read_csv("export.csv")
corpus = corpus_df.reset_index().to_dict(orient="records", )
corpus
```

```python
retriever = retrieve.Encoder(key="index", on=["first_name", "last_name", "email"], normalize=True,
                             encoder=SentenceTransformer("sentence-transformers/all-mpnet-base-v2").encode)
```

```python
retriever.add(corpus)
```

```python
ranker = rank.Encoder(
    key="index",
    on=["first_name", "last_name", "email"],
    encoder=SentenceTransformer(f"sentence-transformers/all-mpnet-base-v2").encode,
)
```

```python
ranker.add(corpus)
```

```python
match = retriever(["will", "mark", "yah"], k=100)
```

```python
 ranker(["will", "mark", "yah"], documents=match, k=30)
```
