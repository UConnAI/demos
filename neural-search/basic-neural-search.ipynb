```python
import torch

from sentence_transformers import SentenceTransformer

embedder = SentenceTransformer("all-MiniLM-L6-v2")

# Corpus with example sentences
corpus = [
    "The sun was setting behind the mountains, casting a golden hue over the valley.",
    "Sarah looked at the old map, tracing the winding river with her finger.",
    "The air smelled of pine and damp earth after the afternoon rain.",
    "A light breeze rustled the leaves, whispering secrets of the forest.",
    "As night fell, the distant howling of wolves echoed through the hills.",
    "The campfire crackled softly, casting flickering shadows on the trees.",
    "John added another log to the fire and glanced at the sky, where stars began to appear.",
    "'I never thought we'd make it this far,' he muttered, staring into the flames.",
    "Sarah smiled, pulling her jacket tighter around her shoulders.",
    "'We still have a long way to go,' she replied, her voice calm and steady.",
    "The wind picked up, carrying the scent of smoke and pine through the clearing.",
    "Far in the distance, the sound of rushing water could be faintly heard.",
    "John leaned back, resting his head against his pack and gazing at the stars.",
    "The fire flickered, its light growing dim as the logs turned to embers.",
    "Sarah stood up and stretched, her breath visible in the cool night air.",
    "'Let's get some rest. Tomorrow will be another long day,' she said softly.",
    "John nodded, pulling his blanket over his shoulders and closing his eyes.",
    "The world around them grew quiet, the forest settling into the deep stillness of night.",
    "Only the distant cry of an owl broke the silence as sleep finally came.",
]

corpus_embeddings = embedder.encode(corpus, convert_to_tensor=True)

# Query sentences:
queries = [
    "The sun was setting behind what natural feature?",
    "Who added a log to the fire and glanced at the sky?",
    "What sound echoed through the hills as night fell?",
]

# Find the closest 5 sentences of the corpus for each query sentence based on cosine similarity
top_k = min(5, len(corpus))
for query in queries:
    query_embedding = embedder.encode(query, convert_to_tensor=True)

    # We use cosine-similarity and torch.topk to find the highest 5 scores
    similarity_scores = embedder.similarity(query_embedding, corpus_embeddings)[0]
    scores, indices = torch.topk(similarity_scores, k=top_k)

    print("\nQuery:", query)
    print("Top 5 most similar sentences in corpus:")

    for score, idx in zip(scores, indices):
        print(corpus[idx], f"(Score: {score:.4f})")
```
